```python
import numpy as np
import pandas as pd

print("🔍 DATA DETECTIVE: Find the Outlier\n")

# Step 1: Generate random data + insert an outlier
np.random.seed(42)
data = list(np.random.randint(40, 60, size=9))  # Normal values
outlier = 120                                    # Deliberate outlier
data.append(outlier)
np.random.shuffle(data)

df = pd.DataFrame({"Values": data})
print(df)

# Step 2: Player guesses the outlier
guess = int(input("\nEnter the value you think is the OUTLIER: "))

# Step 3: IQR-based detection
Q1 = df["Values"].quantile(0.25)
Q3 = df["Values"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

detected_outliers = df[(df["Values"] < lower_bound) | (df["Values"] > upper_bound)]

print("\n📊 IQR Calculation:")
print(f"Q1 = {Q1}, Q3 = {Q3}, IQR = {IQR}")
print(f"Lower Bound = {lower_bound}")
print(f"Upper Bound = {upper_bound}")

# Step 4: Results
print("\n🎯 RESULT:")
if guess in detected_outliers["Values"].values:
    print("✔️ Correct! You successfully detected the outlier like a true Data Detective.")
else:
    print("❌ Not quite. The model flagged the following value(s) as outlier(s):")
    print(detected_outliers["Values"].values)

print("\nGame Over — excellent learning on anomaly detection!")
```
