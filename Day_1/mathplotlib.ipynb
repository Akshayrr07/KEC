```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.arange(1, 11)
y = np.random.randint(10, 100, size=10)
data = np.random.randn(100)

print(x)
print(y)
```

```python
plt.figure(figsize=(6, 4))
plt.plot(x, y, marker='x', linestyle='--', color='blue')
plt.title("Line Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.scatter(x, y, color='green')
plt.title("Scatter Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.bar(x, y, color='orange')
plt.title("Bar Plot")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.barh(x, y, color='purple')
plt.title("Horizontal Bar Plot")
plt.xlabel("Values")
plt.ylabel("Categories")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.hist(data, bins=12, color='brown', edgecolor='black')
plt.title("Histogram")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(6, 6))
plt.pie([10, 20, 30, 40], labels=["A", "B", "C", "D"], autopct='%1.1f%%', startangle=90)
plt.title("Pie Chart")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.boxplot(data)
plt.title("Box Plot")
plt.ylabel("Values")
plt.grid(True)
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(8, 6))

axes[0, 0].plot(x, y, color='blue')
axes[0, 0].set_title("Line Plot")

axes[0, 1].scatter(x, y, color='green')
axes[0, 1].set_title("Scatter Plot")

axes[1, 0].bar(x, y, color='orange')
axes[1, 0].set_title("Bar Plot")

axes[1, 1].hist(data, bins=10, color='red')
axes[1, 1].set_title("Histogram")

plt.tight_layout()
plt.show()
```

```python
matrix = np.random.rand(5, 5)
plt.figure(figsize=(5, 4))
plt.imshow(matrix, cmap='viridis', interpolation='nearest')
plt.colorbar()
plt.title("Heatmap")
plt.show()
```
